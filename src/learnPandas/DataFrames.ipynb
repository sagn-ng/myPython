<p style="color: #3399ff; font-family: 'Times New Roman'">A Pandas <code>DataFrame</code> is a 2D data structure, like a 2D array, or a table with rows and columns.</p>

```python
import pandas as pd

data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}

df = pd.DataFrame(data) #load data into a DataFrame object:
print(type(df))
print(df)
```

Pandas use the <code>loc</code> attribute to return one or more specified row(s).

```python
print(df.loc[0]) #refer to the row index (this e.g returns the first (0_th) row)
```

Passing a single number returns the only row with that index, (return type: <code>Series</code>)
To return many rows, pass in a <code>list</code> '[]' (return type: <code>DataFrames</code>)

```python
print(df.loc[[0, 1]]) #return row 0 and 1
```

Pass a list of names to the <code>index</code> argument to create your own indexes for the rows:

```python
eg = pd.DataFrame(data, index = ["day1", "day2", "day3"])
print(eg)
```

Use the named index in the loc attribute to return the specified row(s).

```python
print(eg.loc["day2"]) #refer to the named index:
```
