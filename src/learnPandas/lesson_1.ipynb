<h2 style="color: white">Creating, Reading and Writing</h2>

<h3 style="color: red">Creating data</h3>
<p>There are two core objects in pandas: the DataFrame and the Series.</p>
<h4 style="color: white">DataFrame</h4>
<p style="font-size: 20px">A <code>DataFrame</code> is a table. It contains an array of individual entries. Each entry has a certain value, corresponding to a row (or record) and a column.<br>
We are using the <code>pd.DataFrame()</code> constructor to generate <code>DataFrame</code> objects.<br>
The syntax for declaring is a <code>dict</code> whose keys are the column names ('Bob' and 'Sue' in this example), and whose values are a list of entries. <p>

```python
import pandas as pd

data = {
  'Bob': ['I liked it.', 'It was awful.'],
  'Sue': ['Pretty good.', 'Bland.']
}

df = pd.DataFrame(data) #load data into a DataFrame object:
print(type(df))
print(df)
```

<p style="font-size: 20px">The <code>dict</code> - <code>list</code> constructor assigns values to the column labels, but just uses an ascending count from 0 (0, 1, 2, 3, ...) for the row labels.<br>But oftentimes we will want to assign these labels ourselves.<br>
The list of row labels used in a DataFrame is known as an <span style="color: green">Index</span>.<br>We can assign values to it by using an <code>index</code> parameter in our constructor:

```python
pd.DataFrame(data, index=['Product A', 'Product B'])
```

<h4 style="color: white">Series</h4>
<p style="font-size: 20px">A <code>Series</code>, by contrast, is a sequence of data values. If a DataFrame is a table, a Series is a list. And in fact you can create one with nothing more than a list:</p>

```python
pd.Series([1, 2, 3, 4, 5])
```

<p style="font-size: 20px">A <code>Series</code> is, in essence, a <span style="color: green">single column of a DataFrame</span>.<br>
So you can assign row labels to the Series the same way as before, using an <code>index</code> parameter.<br>
However, a Series does not have a column name, it only has one overall <code>name</code>:</p>

```python
pd.Series([30, 35, 40], index=['2015 Sales', '2016 Sales', '2017 Sales'], name='Product A')
```

<h4 style="color: red">Reading data files</h3>
<p style="font-size: 20px">A CSV file is a table of values separated by commas ("Comma-Separated Values").<br><br>
Let's see what a real dataset looks like when we read it into a DataFrame. We'll use the <code>pd.read_csv()</code> function to read the data into a DataFrame. This goes thusly:</p>

```python
eg=pd.read_csv('data.csv')
```

<p style="font-size: 20px">We can use the <code>shape</code> attribute to check how large the resulting DataFrame is, for example:</p>

```python
print(eg.shape)
```

<p style="font-size: 20px">So our new DataFrame has 169 records (rows) and 4 columns.<br>
Use the <code>head()</code> or <code>tail()</code> command to grabs the first (or last) 5 rows:</p>

```python
print("The first 5 rows:")
print(eg.head())

print("\nThe last 5 rows:")
print(eg.tail())
```

<p style="font-size: 20px">To make pandas use a column for the index (instead of creating a new one from scratch) when reading from a CSV file, we can specify an <code>index_col</code>.</p>

```python
eg_2=pd.read_csv('data.csv', index_col=1) #make the second col. from the left the index col.
print(eg_2)
```
