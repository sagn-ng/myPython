<p style="font-size: 20px">Big data sets are often stored, or extracted as JSON.
JSON is plain text, but has the format of an object, and is well known in the world of programming, including Pandas.<br>
We use a sample JSON file called <span style="color: #009933"><i>data.json</i></span></p>

```python
import pandas as pd

df = pd.read_json('data.json')

print(df)
```

<p style="font-size: 20px"><span style="color: #009933">Note:</span> also use <code>print(df.to_string())</code> to print the entire DataFrame (like in reading a CSV file).</p>

<span style="color: red">Dictionary as JSON</span>
<p style="font-size: 20px">JSON objects have the same format as Python dictionaries.<br>
If your JSON code is not in a file, but in a Python Dictionary, you can load it into a DataFrame directly:</p>

```python
data = {
  "Duration":{
    "0":60, "1":60, "2":60, "3":45, "4":45, "5":60
  },
  "Pulse":{
    "0":110, "1":117, "2":103, "3":109, "4":117, "5":102
  },
  "Maxpulse":{
    "0":130, "1":145, "2":135, "3":175, "4":148, "5":127
  },
  "Calories":{
    "0":409, "1":479, "2":340, "3":282, "4":406, "5":300
  }
}

df = pd.DataFrame(data)
print(df)
```
