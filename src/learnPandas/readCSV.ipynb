<p style="color: #3399ff; font-family: 'Times New Roman'">A simple way to store big data sets is to use CSV files (comma separated files) - containing plain texts.</p>
<p style="font-size: 20px">We use a sample CSV file called <span style="color: #009933"><i>data.csv</i></span></p>

```python
import pandas as pd

df = pd.read_csv('data.csv') #load the csv file

print(df)
```

<p style="font-size: 20px"><code>print(df)</code> just prints the first 5 rows, then a row with '...' and the last 5 rows.
To print the entire DataFrame, use <code>print(df.to_string())</code> instead.</p>

<code>max_rows</code>
<p style="font-size: 20px">The number of rows returned is defined in Pandas option settings.
You can check your system's maximum rows with the <code>pd.options.display.max_rows</code> statement:</p>

```python
print(pd.options.display.max_rows) 
```

<p style="font-size: 20px">That number in my system is 60, which means if the DataFrame has more than 60 rows, <code>print(df)</code> will only show the headers, and the first and last 5 rows.
However, we can use the statement to increase <code>max_rows</code> to display more data:</p>

```python
pd.options.display.max_rows=200 #for example
```

<p style="font-size: 20px">We can also extract a .csv file from a DataFrame with the <code>to_csv()</code> method of the DaFrame object.<br>That is: <code>pd.DataFrame.to_csv(DataFrame_type, csv_filename)</code></p>
