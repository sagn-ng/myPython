<h2 style="color: white">Renaming and Combining (continue)</h2>

<h4 style="color: purple">join()</h4>

<p style="font-size: 20px"><b style="color: green"><code>join()</code> lets you combine different DataFrame objects which have an index in common.</b>

```python
import pandas as pd

temp1 = pd.DataFrame({
    'City': ['Hanoi', 'HCM']
}, index=['A', 'B'])

temp2 = pd.DataFrame({
    'Population': [8, 9]
}, index=['A', 'B'])

eg_5=temp1.join(temp2) #'temp1' and 'temp2' remain unchanged
print(eg_5)
```

<h4 style="color: purple">merge()</h4>

<p style="font-size: 20px">There are: <b style="color: red">Inner join, Left join, Right join, Outer join</b>, which can be modified by the <code>how</code> argument.</p>

<p style="font-size: 20px"><b style="color: red">Inner join</b>: only keeps records that exist in both dataset:</p>

```python
temp1 = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'Fruits': ['Apple', 'Banana', 'Blueberry', 'Orange'],
    'Score': [9.0, 8.0, 9.5, 9.25],
})
temp2 = pd.DataFrame({
    'id':[1, 2, 5],
    'Sold out': [True, False, True]
})

eg_6=pd.merge(temp1, temp2) #by default: how='inner'
print(eg_6)
```

<p style="font-size: 20px"><b style="color: red">Left (or right) join</b>: keep the left (right) dataset then merge, which might result in <code>NaN</code> cells:</p>

```python
eg_7=pd.merge(temp1, temp2, how='left') #left join
print(eg_7)
```

```python
eg_8=pd.merge(temp1, temp2, how='right') #right join
print(eg_7)
```

<p style="font-size: 20px"><b style="color: red">Outer join</b>: keep both then merge</p>

```python
eg_9=pd.merge(temp1, temp2, how='outer') #outer join
print(eg_9)
```
