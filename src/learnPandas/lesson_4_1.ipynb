<h2 style="color: white">Grouping and Sorting</h2>

<p style="font-size: 20px">Maps allow us to transform data in a DataFrame or Series one value at a time for an entire column. However, often we want to group our data, and then do something specific to the group the data is in.</p>

<h3 style="color: red">Groupwise analysis</h3>

<p style="font-size: 20px">Consider this DataFrame:</p>

```python
import pandas as pd

df = pd.DataFrame({
    'Department': ['IT', 'Finance', 'HR', 'Finance', 'IT'],

    'Gender': ['M', 'M', 'F', 'F', 'M'],

    'Salary': [2500, 1200, 1800, 2000, 1000],

    'Job review': ['great', 'mid', 'mid', 'great', 'low'],

    'Score': [100, 85, 80, 90, 84]
})

print(df)
```

<h4 style="color: white">Basic ideas of groupby()</h4>

<p style="font-size: 20px">We can replicate what <code>value_counts()</code> does by doing the following:</p>

```python
eg_0=df.groupby('Department').Department.count()
print(eg_0)
```

<p style="font-size: 20px"><code>df.groupby('Department').Department.count()</code> returns a <code>Series</code> whose index column contains unique values in <b><i>'Department'</i></b>, and the values are their corresponding frequencies.<br><br>
To get those values, use the simple method <code>.index.to_list()</code>:</p>

```python
temp=eg_0.index.to_list()
print(type(temp))
print(temp)
```

<p style="font-size: 20px">We can use any of the summary functions we've used before with this data.<br><br>
For example, to view the <b>highest</b> salary in different salary values of each department, we do the following:</p>

```python
eg_1=df.groupby('Department').Salary.max()
print(eg_1)
```

<p style="font-size: 20px">You can think of each group generated as a slice of our DataFrame containing only data with values that match. This DataFrame is accessible to us directly using the <code>apply()</code> method, and we can then manipulate the data in any way we want.<br><br>
For example, here's one way of selecting the name of <i style="color: red">the first department to give each kind of review</i> from the dataset:</p>

```python
eg_2=df.groupby('Job review').apply(lambda p: p.Department.iloc[0])
print(eg_2)
```

<h4 style="color: white">Grouping by more than 1 column</h4>

<p style="font-size: 20px">For even more fine-grained control, you can also group by more than 1 column. For an example, here's how we would pick out the <i style="color: red">average salary by job and gender</i>:

```python
eg_3=df.groupby(['Department', 'Gender'])['Salary'].mean()
print(type(eg_3))
print(eg_3)
```
