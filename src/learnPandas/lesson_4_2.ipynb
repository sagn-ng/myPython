<h3 style="color: red">Sorting</h3>

<p style="font-size: 20px">Looking at an example created from the previous DataFrame:</p>

```python
import pandas as pd

df = pd.DataFrame({
    'Department': ['IT', 'Finance', 'HR', 'Finance', 'IT'],

    'Gender': ['M', 'M', 'F', 'F', 'M'],

    'Salary': [2500, 1200, 1800, 2000, 1000],

    'Job review': ['great', 'mid', 'mid', 'great', 'low'],

    'Score': [100, 85, 80, 90, 84]
})

print(df)
```

```python
eg_0=df.groupby(['Department', 'Job review'])['Score'].agg([len, min, max])
eg_0=eg_0.reset_index()

print(eg_0)
```

<p style="font-size: 20px">Notice that <b style="color: green">grouping returns data in index order, not in value order</b>.<br><br>
That is to say, when outputting the result of a <code>groupby</code>, <b style="color: green">the order of the rows is dependent on the values in the index, not in the data</b>.
</p>

To get data in the order want it in we can sort it ourselves. The <code>sort_values()</code> method is handy for this.

```python
eg_1=eg_0.sort_values(by='max')
print(eg_1)
```

<p style="font-size: 20px"><code>sort_values()</code> can also works on multiple columns by passing a <code>list</code> into <code>by=</code>, and it defaults to an <b>ascending sort</b>, where the lowest values go first.<br><br>
However, sometimes we want a descending sort. That goes thusly:</p>

```python
eg_1=eg_0.sort_values(by='max', ascending=False)
print(eg_1)
```

<p style="font-size: 20px">To sort by index values, use the companion method <code>sort_index()</code>. This method has the same arguments and default order:</p>

```python
eg_2=eg_0.sort_index()
print(eg_2)
```
