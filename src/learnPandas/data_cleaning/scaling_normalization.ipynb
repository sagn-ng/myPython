<h3 style="color: white">Scaling and Normalization</h3>

<h4 style="color: red">Get the environment set up</h4>

```python
# modules we'll use
import pandas as pd
import numpy as np

# for Box-Cox Transformation
from scipy import stats

# for min_max scaling
from mlxtend.preprocessing import minmax_scaling

# plotting modules
import seaborn as sns
import matplotlib.pyplot as plt
```

<h4 style="color: red">Scaling vs Normalization</h4>

<p style="font-size: 20px">In both cases, you're transforming the values of numeric variables so that the transformed data points have specific helpful properties. The difference is that:<br>
<ul style="color: green">
<li>in scaling, you're changing the range of your data, while</li>
<li>in normalization, you're changing the shape of the distribution of your data.</li>
</ul></p>

<h4 style="color: red">Scaling</h4>

<p style="font-size: 20px">This means that you're transforming your data so that it fits within a specific scale, like 0-100 or 0-1.<br><br>
For example, you might be looking at the prices of some products in both Yen and US Dollars. One US Dollar is worth about 100 Yen, but if you don't scale your prices, methods like SVM or KNN will consider a difference in price of 1 Yen as important as a difference of 1 US Dollar! This clearly doesn't fit with our intuitions of the world.<br><br>
And what about you're looking at something like height and weight? It's not entirely clear how many pounds should equal one inch (or how many kilograms should equal one meter).</p>

<p style="font-size: 20px">We'll be looking at the made-up example below. (There will be exercises for better visualization)</p>

```python
# generate 1000 data points randomly drawn from an exponential distribution
original_data = np.random.exponential(size=1000)

# mix-max scale the data between 0 and 1
scaled_data = minmax_scaling(original_data, columns=[0])

# plot both together to compare
fig, ax = plt.subplots(1, 2, figsize=(15, 3))
sns.histplot(original_data, ax=ax[0], kde=True, legend=False)
ax[0].set_title("Original Data")
sns.histplot(scaled_data, ax=ax[1], kde=True, legend=False)
ax[1].set_title("Scaled data")
plt.show()
```

<p style="font-size: 20px">Notice that the shape of the data doesn't change, but that instead of ranging from 0 to 8ish, it now ranges from 0 to 1.</p>

<h4 style="color: red">Normalization</h4>

<p style="font-size: 20px">Scaling just changes the range of your data.<br><br>
Normalization is a more radical transformation. <b style="color: green">The point of normalization is to change your observations so that they can be described as a normal distribution.</b></p>

<p style="font-size: 20px">The method we're using to normalize here is called the <i style="color: red">Box-Cox Transformation.</i> Let's take a quick peek at what normalizing some data looks like:</p>

```python
# normalize the exponential data with boxcox
normalized_data = stats.boxcox(original_data)

# plot both together to compare
fig, ax=plt.subplots(1, 2, figsize=(15, 3))
sns.histplot(original_data, ax=ax[0], kde=True, legend=False)
ax[0].set_title("Original Data")
sns.histplot(normalized_data[0], ax=ax[1], kde=True, legend=False)
ax[1].set_title("Normalized data")
plt.show()
```
