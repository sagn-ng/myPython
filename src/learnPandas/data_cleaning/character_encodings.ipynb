<h4 style="color: red">What are encodings</h4>

<p style="font-size: 20px"><b style="color: green">Character encodings</b> are specific sets of rules for <b style="color: green">mapping from raw binary byte strings to characters that make up human-readable text</b> (like "hi").<br><br>
There are many different encodings, and if you tried to read in text with a different encoding than the one it was originally written in, you ended up with scrambled text called <b style="color: red">"mojibake"</b> (said like mo-gee-bah-kay).</p>

<details>
<summary style="font-size: 20px">There are lots of different character encodings, but the main one you need to know is UTF-8.</summary>
<p style="font-size: 20px">UTF-8 is the standard text encoding. All Python code is in UTF-8 and, ideally, all your data should be as well. It's when things aren't in UTF-8 that you run into trouble.</p>
</details>

<p style="font-size: 20px">There are two main data types you'll encounter when working with text in Python 3. One is is the string (<code>str</code>), which is what text is by default:</p>

```python
# start with a string
before = "This is the euro symbol: €"

# check to see what datatype it is
type(before)
```

<p style="font-size: 20px">The other data is the <code>bytes</code> type, i.e <b>a sequence of integers.</b> You can convert a string into bytes by specifying which encoding it's in:</p>

```python
# encode it to a different encoding, replacing characters that raise errors
after = before.encode("utf-8", errors="replace")

# check the type
type(after)
```

<p style="font-size: 20px">If you look at a <code>bytes</code> object, you'll see that it has a <span style="color: red">b</span> in front of it.<br><br>
That's because <b style="color: green">bytes are printed out as if they were characters encoded in ASCII.</b>.<br>Here you can see that our euro symbol has been replaced with some mojibake that looks like <span style="color: red">"\xe2\x82\xac"</span> when it's printed like an ASCII string.</p>

```python
# take a look at what the bytes look like
after
```

<p style="font-size: 20px">When we convert our bytes back to a string with the correct encoding, we can see that our text is all there correctly!</p>

```python
print(after.decode(encoding='utf-8'))
```

<details>
<summary style="color: red">Note</summary>
<p style="font-size: 20px">However, when we try to use a different encoding to map our bytes into a string, we get an error. This is because the encoding we're trying to use doesn't know what to do with the bytes we're trying to pass it.</p>
<p style="font-size: 20px">We can also run into trouble if using the wrong encoding to map from a string to bytes.</p>
<p style="font-size: 20px">For example, if we try to convert a string to bytes for ASCII using <code>encode()</code>, nothing would happen if the text is in ASCII. But since our text isn't in ASCII, there will be some characters it can't handle. We can automatically replace those. If so, however, any characters not in ASCII will just be replaced with the unknown character. Then, when we convert back to a string, the character will be replaced with the unknown character. The dangerous part about this is that there's not a way to tell which character it should have been.</p>
</details>

```python
# start with a string
before = "This is the euro symbol: €"

# encode it to a different encoding, replacing characters that raise errors
after = before.encode("ascii", errors = "replace")

# convert it back to utf-8
print(after.decode("ascii"))

# We've lost the original underlying byte string! It's been 
# replaced with the underlying byte string for the unknown character
```

<h4 style="color: red">Reading files with encoding problems</h4>

<p style="font-size: 20px">Suppose we want to apply an encoding on a <code>.csv</code> file. We add the argument: <code>encoding=</code> inside <code>pd.read_csv()</code> to specify that encoding. <b style="color: green">(<code>encoding='utf-8'</code> by default).</b></p>

<p style="font-size: 20px">Most files you'll encounter will probably be encoded with UTF-8. This is what Python expects by default, so most of the time you won't run into problems.</p>

```python
# look at the first ten thousand bytes to guess the character encoding
with open('catalog.csv', 'rb') as rawdata:
    result = charset_normalizer.detect(rawdata.read(10000))

# check what the character encoding might be
print(result)
```

<p style="font-size: 20px">Now, using <code>charset_normalizer</code>, we know that the encoding is probably UTF-8, so we can open it easily with:</p>

```python
# read in the file with the encoding detected by charset_normalizer
ourData = pd.read_csv('catalog.csv')

# look at the first few lines
ourData.head()
```
