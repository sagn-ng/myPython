<h3 style="color: white">Parsing Dates</h3>

<h4 style="color: red">Get our environment set up</h4>

<p style="font-size: 20px">We'll be working with a dataset that contains information on landslides that occured between 2007 and 2016.</p>

```python
# modules we'll use
import pandas as pd
import numpy as np
import seaborn as sns
import datetime

# read in our data
landslides = pd.read_csv("catalog.csv")
```

<h4 style="color: red">Check the data type of our date column</h4>

<p style="font-size: 20px">We begin by taking a look at the first five rows of the data.</p>

```python
landslides.head()
```

<p style="font-size: 20px">We'll be working with the "date" column from the <code>landslides</code> dataframe. Let's make sure it actually looks like it contains dates.</p>

```python
# print the first few rows of the date column
print(landslides['date'].head())
```

<p style="font-size: 20px">Those are dates! But it doesn't mean that Python knows that they're dates.<br>
Notice that at the bottom of the output of <code>head()</code>, you can see that it says that the data type of this column is <code>str</code> rather than <code>datetime64</code>.<br>
We can also look at just the <code>dtype</code> of a column without printing the first few rows:</p>

```python
# check the data type of our date column
landslides['date'].dtype
```

<h4 style="color: red">Convert our date columns to datetime</h4>

<p style="font-size: 20px">It's time to convert our "date" column so that it is recognized as a date. <b style="color: green">This is called "parsing dates" because we're taking in a string and identifying its component parts.</b>
</p>

<p style="font-size: 20px">The basic idea is that you need to point out which parts of the date are where and what punctuation is between them.<br>
There are lots of possible parts of a date, but the most common are <b style="color: green"><code>%d</code> for day, <code>%m</code> for month, <code>%y</code> for a two-digit year and <code>%Y</code> for a four digit year.</b><br><br>
Some examples:
<ul>
<li>1/17/07 has the format "%m/%d/%y"</li>
<li>17-1-2007 has the format "%d-%m-%Y"</li>
</ul>
Looking back up at the head of the "date" column in the <code>landslides</code> dataset, we can see that <span style="color: green">it's in the format "month/day/two-digit year"</span>, so we can use the same syntax as the first example to parse in our dates:
</p>

```python
# create a new column, date_parsed, with the parsed dates
landslides['date_parsed'] = pd.to_datetime(landslides['date'], format="%m/%d/%y")
```

<p style="font-size: 20px">Now when I check the first few rows of the new column, I can see that the <code>dtype</code> is <code>datetime64</code>. I can also see that my dates have been slightly rearranged so that they fit the default order <code>datetime</code> objects <span style="color: green">(year-month-day).</span>
</p>


```python
# print the first few rows
landslides['date_parsed'].head()
```

<p style="font-size: 20px"><span style="color: red">Note:</span> When meeting multiple data formats in the same column, you can add the argument: <code>infer_datetime_format=True</code> to make Python guess the format.<br>
But overall, it's not recommended to have multiple formats in one column.</p>
