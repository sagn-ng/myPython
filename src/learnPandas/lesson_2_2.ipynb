<h3 style="color: red">Manipulate the index</h3>
<p style="font-size: 20px">Label-based selection derives its power from the labels in the index. Critically, the index we use is not immutable. We can manipulate the index in any way we see fit.<br>
<br>
The <code>set_index()</code> method can be used to do the job.<br>
<i>(This is useful if you can come up with an index for the dataset which is better than the current one.)</i>
<br><br>
For example, consider a newly created DataFrame below, named <code>new_df</code>.
<br><br>
Here is what happens when we <code>set_index</code> to the <code>name</code> field:

```python
import pandas as pd

new_df=pd.DataFrame({
    'price': [15, 10, 20, 25],
    'name': ['Apple', 'Banana', 'Orange', 'Blueberry'],
    'discount (%)': [10, 7, 10, 25],
    'remaining quantity': [1200, 1000, 2500, 1000]
})

print(new_df.set_index('name'))
```

<h3 style="color: red">Conditional selection</h3>
<p style="font-size: 20px">To do interesting things with the data, however, we often need to ask questions based on conditions.<br>
For example, suppose that we're interested in fruits whose prices are >12.<br>
We start checking if each fruit's price is bigger than 12 or not:<br>
</p>

```python
x=(new_df.price>12)
print(type(x))
print(x)
```

<p style="font-size: 20px">This operation produced a <code>Series</code> of <code>True/False</code> booleans based on the <code>price</code> of each record. This result can then be used inside of <code>loc</code> to select the relevant data:</p>

```python
eg_2_0=new_df.loc[x] #or: new_df.loc[new_df.price > 12]
print(type(eg_2_0))
print(eg_2_0) #eg_2_0is a DataFrame
```

<p style="font-size: 20px">We can use some operators like <code>&</code> (and), <code>|</code> (or) to combine many conditions for filtering.</p>

```python
eg_2_1=new_df.loc[(new_df.price>12) & (new_df['discount (%)']>10)]
print(eg_2_1)
```

```python
eg_2_2=new_df.loc[(new_df.price>12) | (new_df['remaining quantity']>1000)]
print(eg_2_2)
```

<p style="font-size: 20px"><code>isin</code> is lets you select data whose value "is in" a list of values.<br>
For example, here's how we can use it to select only <code>Apple</code> and <code>Orange</code>:</p>

```python
eg_2_3=new_df.loc[new_df.name.isin(['Apple', 'Orange'])]
print(eg_2_3)
```

<h3 style="color: red">Assigning data</h3>
<p style="font-size: 20px">Going the other way, assigning data to a DataFrame is easy. You can assign either a constant value:</p>

```python
new_df['sold out']=False #it's like adding a new column
print(new_df['sold out'])
```

<p style="font-size: 20px">or with an iterable of values:</p>

```python
new_df['rank']=[1, 3, 2, 4]
print(new_df)
```
