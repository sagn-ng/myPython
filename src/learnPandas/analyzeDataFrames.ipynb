<span style="color: red">Viewing the data</span>
<p style="font-size: 20px">One of the most used method for getting a quick overview of the DataFrame, is the <code>head()</code> method - it returns the headers and a specified number of rows, starting from the top.</p>

```python
import pandas as pd

df=pd.read_csv('data.csv')

print(df.head(10))
#if the number of rows is not specified in head(), the method will return the first 5 rows
```

<p style="font-size: 20px">There is also a <code>tail()</code> method for viewing the last rows of the DataFrame.<br>
E.g - viewing the last 5 rows of the DataFrame:</p>

```python
print(df.tail())
```

<span style="color: red">Info about the data</span>
<p style="font-size: 20px">The DataFrames object has the <code>info()</code> method that gives us more information about the data set.</p>

```python
print(df.info())
```

<p> The <span style="color: #ffffff">"Calories"</span> column has 5 rows with no values, i.e <span style="color: #009933"><i>empty/null values</i></span>. Those null values can be bad when analyzing data, and you should consider removing rows with empty values. This is a step towards what is called <span style="color: #009933"><i>cleaning data</i></span>.
