<h2 style="color: white">Renaming and Combining</h2>

<h3 style="color: red">Renaming</h3>

<p style="font-size: 20px">Consider the following DataFrame:</p>

```python
import pandas as pd

df = pd.DataFrame({
    'Department': ['IT', 'Finance', 'HR', 'Finance', 'IT'],

    'Gender': ['M', 'M', 'F', 'F', 'M'],

    'Salary': [2500, 1200, 1800, 2000, 1000],

    'Job review': ['great', 'mid', 'mid', 'great', 'low'],

    'Score': [100, 85, 80, 90, 84]
})

print(df)
```

<p style="font-size: 20px">The first function we'll introduce here is <code>rename()</code>, which <b style="color: green">lets you change index names and/or column names.</b><br><br>
For example, to change the <code>Score</code> column in our dataset to <code>points</code>, we would do:</p>

```python
eg_0=df.rename(columns={'Score': 'points'})
print(eg_0)
```

<p style="font-size: 20px"><code>rename()</code> <b style="color: green">lets you rename index or column values by specifying a <code>index</code> or <code>column</code> keyword parameter,</b> respectively. It supports a variety of input formats, but usually a Python dictionary is the most convenient.<br><br>
Here is an example:</p>

```python
#rename the 0_th and 1_th indexed rows as 'firstEntry' and 'secondEntry'
eg_1=df.rename(index={0: 'firstEntry', 1: 'secondEntry'})
print(eg_1)
```

<p style="font-size: 20px">Renaming columns is probably very often, but renaming index values is very rare (since <code>set_index()</code> is usually more convenient).<br><br>
<b style="color: green">Both the row index and the column index can have their own <code>name</code> attribute.</b><br><br>
The complimentary <code>rename_axis()</code> method may be used to change these names. For example:</p>

```python
#rename the index column as 'numbers' and add the name 'fields' for columns' names:
eg_2=df.rename_axis('numbers', axis='index').rename_axis('fields', axis='columns')
print(eg_2)
```

<h3 style="color: red">Combining</h3>

<p style="font-size: 20px">Pandas has three core methods for combing DataFrame/Series. In order of increasing complexity, these are <code>concat()</code>, <code>join()</code>, and <code>merge()</code>.</p>

<h4 style="color: purple">concat()</h4>

<p style="font-size: 20px">Given a list of elements, <b style="color: green">this function will smush those elements together along an axis</b> (<code>axis=0</code> by default).</p>

<p style="font-size: 20px"><span style="color: orange">Combine by rows</span> (data in different DataFrames/Series objects but having the name fields/columns):</p>

```python
temp1 = pd.DataFrame({'Name': ['A', 'B']})
temp2 = pd.DataFrame({'Name': ['C', 'D']})

eg_3 = pd.concat([temp1, temp2]) #axis=0

print(eg_3)
```

<p style="font-size: 20px">We see in the result that, indexes are <code>0 1 0 1</code>, we can reset the index to fix this by adding: <code>ignore_index=True</code>.</p>

<p style="font-size: 20px"><span style="color: orange">Combine by columns</span> (it's like adding new columns):</p>

```python
temp1 = pd.DataFrame({
    'Fruits': ['Apple', 'Banana', 'Blueberry', 'Orange'],
    'Score': [9.0, 8.0, 9.5, 9.25],
})
temp2 = pd.DataFrame({
    'Price': [10000, 7000, 12000, 11000],
    'Sold out': [True, False, False, True]
})

eg_4=pd.concat([temp1, temp2], axis=1) #or axis='columns'
print(eg_4)
```
