<h2 style="color: white">Indexing, Selecting & Assigning</h2>

<h3 style="color: red">Native accessors</h3>
<p style="font-size: 20px">Consider the DataFrame <code>review</code></p>

```python
import pandas as pd

data = {
  "calories": [420, 380, 390, 250],
  "duration": [50, 40, 45, 10]
}

reviews = pd.DataFrame(data)
```

<p style="font-size: 20px">A <code>book</code> object, for example, might have a <code>title</code> property, which we can access by calling <code>book.title</code>.<br>
Columns in a pandas DataFrame work in much the same way.<br>
Hence to access the <code>calories</code> column of <code>reviews</code> we can use:</p>

```python
eg_0=reviews.calories #eg_0 is a Series
print(type(eg_0))
print(eg_0)
```

<p style="font-size: 20px">Or by using the indexing (<code>[]</code>) operator like a <code>dict</code>:</p>

```python
eg_0=reviews['calories']
print(eg_0)
```

<p style="font-size: 20px">These are the 2 ways of selecting a specific Series (or column) out of a DataFrame.<br><br>
<b style="color: green">Note: </b>However, <code>[]</code> has the advantage that it can handle column names with reserved characters (e.g: if we had a <code>person name</code> column, <code>reviews.person name</code> wouldn't work).</p>
<p style="font-size: 20px">To drill down to a specific cell, use the indexing operator <code>[]</code> once more:</p>

```python
print(reviews['calories'][2]) #row number 2 of the 'calories' column
```

<h3 style="color: red">Indexing in pandas</h3>
<p style="font-size: 20px">pandas has its own accessor operators, <code>loc</code> and <code>iloc</code>, which are the ones you're supposed to be using for more advanced operations.<br><br>
Both <code>loc</code> and <code>iloc</code> are <b style="color: green">row-first, column second</b>, opposite of what we do in native accessors.</p>

<h4 style="color: white">Index-based selection</h4>
<p style="font-size: 20px">This is: <b style="color: green">selecting data based on its numerical position</b> in the data.<br>
<code>iloc</code> follows this paradigm. E.g, to select the first row in a DataFrame:</p>

```python
eg_1=reviews.iloc[0] #eg_1 is also a Series
print(type(eg_1))
print(eg_1)
```

<p style="font-size: 20px">We can extract specific rows/columns by passing lists inside the <code>iloc[]</code>.<br><br>
For example, to get column 0 and 1, row 0 and 3 in <code>reviews</code>, we would do:</p>

```python
eg_2=reviews.iloc[[0, 3], [0, 1]] #eg_2 is a DataFrame
print(type(eg_2))
print(eg_2)
```

<h4 style="color: white">Label-based selection</h4>
<p style="font-size: 20px">This is: <b style="color: green">selecting data based on its labeled position</b> in the data.<br>
Things work almost the same as in <code>iloc</code>, just now columns are accessed by labels, not indexes.<br>

For example:</p>

```python
eg_3=reviews.loc[[0, 3], ['calories', 'duration']] #it yields the same result
print(eg_3)
```

<h4 style="color: green">*Note</h4>
<p style="font-size: 20px">When passing a lists of number indexes for rows (or columns), say <code>[a, b]</code>, <code>iloc[]</code> treat it as <code>a, a+1,..., b-1</code> (b is exclusive); but in <code>loc[]</code>, b is inclusive.</p>
