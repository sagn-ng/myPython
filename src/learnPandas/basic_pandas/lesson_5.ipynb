<h2 style="color: white">Data Types and Missing Values</h2>

<h3 style="color: red">Dtype</h3>

<p style="font-size: 20px">The data type for a column in a DataFrame or a Series is known as the dtype.<br><br>
You can use the <code>dtype</code> property to <b style="color: green">grab the type of a specific column (i.e a <code>Series</code>).</b> For instance, we can get the dtype of the <code>Score</code> column in the <code>df</code> DataFrame:</p>

```python
import pandas as pd

df = pd.DataFrame({
    'Department': ['IT', 'Finance', 'HR', 'Finance', 'IT'],

    'Gender': ['M', 'M', 'F', 'F', 'M'],

    'Salary': [2500, 1200, 1800, 2000, 1000],

    'Job review': ['great', 'mid', 'mid', 'great', 'low'],

    'Score': [100, 85, 80, 90, 84]
})
```

```python
print(df.Score.dtype)
```

<p style="font-size: 20px">Alternatively, the <b style="color: green"><code>dtypes</code> property returns the <code>dtype</code> of every column in a DataFrame:</b></p>

```python
print(df.dtypes)
```

<p style="font-size: 20px">Data types tell us something about how pandas is storing the data internally. <code>int64</code> means that it's using a 64-bit integer; <code>float64</code> means a similarly sized floating number instead, and so on.<br><br>
(sometimes) <b style="color: green">columns consisting entirely of strings do not get their own type; they are instead given the object type.</b><br><br>
It's possible to convert a column of one type into another wherever such a conversion makes sense by using the <code>astype()</code> function. For example, we may transform the <code>Score</code> column from its existing <code>int64</code> data type into a </code>float64</code> one:

```python
eg_0=df.Score.astype('float64')
print(eg_0)
```

<p style="font-size: 20px"><b style="color: green">A DataFrame or Series index also has its own <code>dtype</code</p>

```python
print(df.index.dtype)
```

<h3 style="color: red">Missing data</h3>

<p style="font-size: 20px">We add this extra column below for visualization:</p>

```python
df['Description']=['intelligent', 'hardworking', None, 'caring', None]
print(df)
```

<p style="font-size: 20px">Entries missing values are given the value <code>NaN</code>, ("Not a Number"), and these <code>NaN</code> values are always of the <code>float64</code> dtype.<br><br>
To select NaN entries you can use <code>pd.isnull()</code> (or its companion <code>pd.notnull()</code>). This is meant to be used thusly:</p>

```python
eg_1=df[pd.isnull(df['Description'])] #eg_1 is a DataFrame
print(eg_1)
```

<p style="font-size: 20px">Pandas provides a really handy method to <b style="color: green">replace mising values:</b> <code>fillna()</code>.<br>
The method provides a few different strategies for mitigating such data. For example, we can simply replace each <code>NaN</code> with an <code>"Unknown"</code>:</p>

```python
eg_2=df['Description'].fillna("Unknown") #eg_2 is a Series
print(eg_2)
```

<p style="font-size: 20px">Alternatively, we may have a non-null value that we would like to replace.<br><br>
For example, all <span style="color: green">'low'</span> job reviews has changed into <span style="color: green">'mid'</span>.<br>
We reflect this in the dataset by using the <code>replace</code> method:</p>

```python
eg_3=df['Job review'].replace('low', 'mid') #this returns a Series
print(eg_3)
```

<p style="font-size: 20px">The <code>replace()</code> method is worth mentioning here because it's <b style="color: green">handy for replacing missing data which is given some kind of sentinel value in the dataset: things like</b> <i style="color: red">"Unknown", "Undisclosed", "Invalid"</i>, and so on.</p>
