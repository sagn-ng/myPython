<h2 style="color: white">Summary Functions and Maps</h2>

<p style="font-size: 20px">We'll be using the following data for demonstration:</p>

```python
import pandas as pd

df=pd.DataFrame({
    'name': ['Apple', 'Banana', 'Orange', 'Blueberry'],
    'price': [15, 10, 20, 25],
    'discount (%)': [10, 7, 10, 25],
    'remaining quantity': [1200, 1000, 2500, 1000],
    'rank': [1, 3, 2, 4]
})

print(df)
```

<h3 style="color: red">Summary functions</h3>
<p style="font-size: 20px">Pandas provides many simple "summary functions" (not an official name) which restructure the data in some useful way.<br>
For example, consider the <code>describe()</code> method:</p>

```python
df.price.describe()
```

<p style="font-size: 20px">Note that the <code>describe()</code> method is type-aware, which means that its output changes based on the input's data type<br>
For example, for the string data type, we get:</p>

```python
df.name.describe()
```

<p style="font-size: 20px">If you want to get some particular summary statistic about a column in a DataFrame or a Series, there is a helpful function.<br>
<br>
For example, to see the mean of the <code>price</code> allotted, we can use the <code>mean()</code> function:</p>

```python
df.price.mean()
```

<p style="font-size: 20px">In addition, to see the median value, use <code>median()</code>. To see a list of unique values we can use the <code>unique()</code> function, or the <code>value_counts()</code> to also see how often those values occur in the dataset.</p>

<h3 style="color: red">Maps</h3>
<p style="font-size: 20px"> In data science we often need to create new representations from existing data, or for transforming data into the format that we want later.<br>
Maps are what handle this work, making them extremely important!</p>

<h4><li style="color: white"><code>map()</code></li></h4>

<p style="font-size: 20px"><code>map()</code> is the first, and slightly simpler one, <b style="color: green">used for Series.</b><br>
It applies the transformation on every element of the Series.<br>
However, it <b style="color: green">creates a new Series and leaves the original one unchanged.</b></p>

```python
s = pd.Series([2, 5, 10])

eg_0=s.map(lambda x: x*2)
print(eg_0)
```

<p style="font-size: 20px">Now let's try to use <code>map()</code> on a whole DataFrame, i.e only <b style="color: green">modifying columns.</b><br>
<br>
For example, suppose that we wanted to <i style="color: green">remean</i> the <code>price</code> of the fruits to 0. We can do this as follows:</p>

```python
mean_price=df.price.mean()
eg_1=df.price.map(lambda x: x-mean_price)
print(type(eg_1))
print(eg_1)
#i.e: replace all values X in 'price' with X - avg(price), leaving 'df' unchanged.
```

<h4><li style="color: white"><code>apply()</code></li></h4>

<p style="font-size: 20px"><code>apply()</code> is more flexible since it can be used for  <b style="color: green">transforming single elements, rows, column</b><br>


<p style="font-size: 20px">The <code>apply()</code> has the <code>axis</code> argument that we should be aware of.<br>
It can be:<br>
<ul style="font-size: 20px">
<li><b style="color: green">0 ('index')</b> (by default), meaning the transformation is applied on each <b style="color: green">column</b>;</li>
<li><b style="color: green">1 ('columns')</b>, meaning the transformation is applied on each <b style="color: green">row</b> (this is what <code>map</code> cannot do).</li>
</ul>
</p>

```python
eg_2=df.apply(lambda row: row.price-mean_price, axis=1)
#we're modifying the 'price' attribute of each record (row)
print(eg_2)
```

<p style="font-size: 20px"><b style="color: red">Note that:
<ul style="font-size: 20px">
<li></b>using <code>apply()</code> for a Series doesn't involve the <code>axis</code> argument;</li>
<li>it and using in a DataFrame with <code>axis=1</code> are exactly the same as <code>map()</code>.</li>
</ul>
</p>

```python
eg_3=df.price.apply(lambda p: p-mean_price)
print(eg_3) #it yields the same result as eg_2
```

<p style="font-size: 20px">Now let's try to apply calculations on rows with <code>apply()</code>, which is a thing <code>map()</code> cannot do.<br>
For example, we want to make a new <code>Series</code> to see the final prices after discounts of the fruits:</p>

```python
eg_4=df.apply(
    lambda row: row['price']-row['price']*row['discount (%)']/100,
    axis=1)
print(eg_4)
```

<p style="font-size: 20px">And we might want to add this Series to the dataset (i'll make a copy to keep the original one):</p>

```python
eg_5=df.copy()
eg_5['price after discount']=eg_4
print(eg_5)
```

<h4 style="color: red">Vectorization</h4>

<p style="font-size: 20px">Consider the example of calculating prices above (eg_4).<br>
<br>
There is a faster way (i'll also make a copy):</p>

```python
eg_6=df.price
eg_6-=eg_6*df['discount (%)']/100 #directly apply the transformation
print(eg_6)
```
